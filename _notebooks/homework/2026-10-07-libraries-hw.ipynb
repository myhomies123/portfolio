# 3.14 Libraries Homework

**MCQ: 4/4.** I used a standard-library module to organize and summarize repeated activity data.

```python
# CODE_RUNNER: Homework - library data summary
from collections import Counter
activities = ['CyberPatriot', 'Speech', 'CyberPatriot', 'Coding']
counts = Counter(activities)
for activity, count in counts.items(): print(activity, count)
assert counts['CyberPatriot'] == 2
print('Most frequent:', counts.most_common(1)[0][0])
```

## Evidence
The imported `Counter` library replaces a longer manual counting loop and the assertion verifies its result.
