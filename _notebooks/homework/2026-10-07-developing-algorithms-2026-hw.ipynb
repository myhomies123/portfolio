**MCQ: 4/4.** This version develops a repeatable search algorithm and verifies normal and no-match cases.

```python
# CODE_RUNNER: Homework - develop and test a search algorithm
def find_by_spec(records, spec):
    for record in records:
        if record['spec'] == spec: return record
    return None
records = [{'spec':'1.1','name':'Flywheel'}, {'spec':'2.1','name':'Clutch'}]
print(find_by_spec(records, '2.1'))
print(find_by_spec(records, '9.9'))
assert find_by_spec(records, '2.1')['name'] == 'Clutch'
assert find_by_spec(records, '9.9') is None
```
