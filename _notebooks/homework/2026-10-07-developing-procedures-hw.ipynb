# 3.13 Developing Procedures Homework

**MCQ: 4/4.** I wrote procedures with parameters and return values, then reused them in multiple tests.

```python
# CODE_RUNNER: Homework - reusable score procedures
def points_for_level(stars, bonus):
    return stars * 10 + bonus
def progress_message(player, stars, bonus):
    return f'{player} earned {points_for_level(stars, bonus)} points.'
print(progress_message('Rohan', 4, 5))
print(progress_message('Teammate', 2, 0))
assert points_for_level(4, 5) == 45
```

## Evidence
Changing the procedure inputs changes the returned score without duplicating the calculation.
