This notebook completes both Popcorn Hacks and the Recovery Center Cost Report using variables and evaluated Python math expressions.

## Popcorn Hack 1 — Basic Arithmetic Operators

```python
# CODE_RUNNER: Popcorn Hack 1 - Basic Arithmetic Operators
value_a = 100
value_b = 10
value_c = 25

addition = value_a + value_b + value_c
subtraction = value_a - value_b
multiplication = value_b * value_c
division = value_a / value_b

print("Total:", addition)
print("Difference:", subtraction)
print("Product:", multiplication)
print("Division:", division)

assert (addition, subtraction, multiplication, division) == (135, 90, 250, 10.0)
print("Validation: all four arithmetic expressions are correct.")
```

**Verified results**

- Total: 135
- Difference: 90
- Product: 250
- Division: 10.0
- Validation: all four arithmetic expressions are correct.

## Popcorn Hack 2 — Recovery Center Session Costs

```python
# CODE_RUNNER: Popcorn Hack 2 - Recovery Center weekly and monthly cost
cost_per_session = 35
sessions_per_week = 3
weeks_per_month = 4

weekly_cost = cost_per_session * sessions_per_week
monthly_cost = weekly_cost * weeks_per_month

print("Weekly Cost: $", weekly_cost, sep="")
print("Monthly Cost: $", monthly_cost, sep="")

assert weekly_cost == 105
assert monthly_cost == 420
print("Validation: weekly and monthly costs are correct.")
```

## Homework — Recovery Center Cost Report

The report keeps every input in a variable, so changing a session price or schedule automatically updates every result.

```python
# CODE_RUNNER: HOMEWORK - Recovery Center Cost Report
# Given values
cost_per_session = 35
sessions_per_week = 3
weeks_per_month = 4
enrollment_fee = 75

# 1. Weekly cost: session price times sessions each week
weekly_cost = cost_per_session * sessions_per_week

# 2. Monthly cost: weekly cost times weeks each month
monthly_cost = weekly_cost * weeks_per_month

# 3. Total first-month cost: monthly cost plus one-time enrollment fee
total_cost_with_fee = monthly_cost + enrollment_fee

# 4. Average daily cost: monthly cost divided by 30 days
average_daily_cost = monthly_cost / 30

print("Weekly Cost: $", weekly_cost, sep="")
print("Monthly Cost: $", monthly_cost, sep="")
print("Total Cost with Fee: $", total_cost_with_fee, sep="")
print("Average Daily Cost: $", average_daily_cost, sep="")

assert weekly_cost == 105
assert monthly_cost == 420
assert total_cost_with_fee == 495
assert average_daily_cost == 14.0
print("Validation: all required cost calculations are correct.")
```

**Verified results**

- Weekly Cost: $105
- Monthly Cost: $420
- Total Cost with Fee: $495
- Average Daily Cost: $14.0
- Validation: all required cost calculations are correct.
