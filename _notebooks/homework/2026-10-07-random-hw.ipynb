# 3.15 Randomness Homework

**MCQ: 4/4.** I generated valid IDs, selected complete records, and tested repeatable bounds.

```python
# CODE_RUNNER: Homework - random QA simulator
import random
records = [{'spec':'1.1','name':'Flywheel'}, {'spec':'2.1','name':'Clutch'}]
def qa_run():
    record = random.choice(records)
    action = random.choice(['GET', 'POST', 'PUT', 'DELETE'])
    test_id = random.randint(1000, 9999)
    return test_id, action, record
for _ in range(3):
    test_id, action, record = qa_run(); print(test_id, action, record['spec'])
assert all(1000 <= qa_run()[0] <= 9999 for _ in range(20))
```

## Evidence
The tests confirm every random ID stays in range, while repeated IDs or actions remain expected random behavior.
