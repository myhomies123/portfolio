### Popcorn Hack 1 — Repair the silent failure (0.10)

Original prediction: amount 0 prints nothing; this is a logic bug, not a Python exception. The fix adds the missing answer. The threshold is strictly greater than 100, so 100 belongs to the low-amount branch.


````python
# CODE_RUNNER: 3.6 Popcorn 1 - Both outcomes print
for amount_requested in [0, 100, 101]:
    print("Amount:", amount_requested)
    # 1 condition -> 2 outcomes -> 2 answers
    if amount_requested > 100:
        print("This message is asking for a large amount of money.")
    else:
        print("This message is not asking for a large amount of money.")
````

### Popcorn Hack 2 — Fix unreachable branches (0.10)

Original prediction: score 9 incorrectly prints “Be a little careful, but this is probably fine.” The original `>= 6` and `>= 9` branches can never run because the earlier `>= 3` consumes all those scores. Testing the highest thresholds first makes all four answers reachable.


````python
# CODE_RUNNER: 3.6 Popcorn 2 - Fix the threshold order
for risk_score in [0, 3, 6, 9]:
    print("Risk score:", risk_score)
    # 3 thresholds -> 4 outcomes -> 4 answers
    if risk_score >= 9:
        print("Do not reply. This is almost certainly a scam.")
    elif risk_score >= 6:
        print("This looks risky. Check with someone you trust before replying.")
    elif risk_score >= 3:
        print("Be a little careful, but this is probably fine.")
    else:
        print("Nothing suspicious in this message.")
````

### Homework Hack — Score a message, then advise (0.80)

Signals follow the lesson exactly: **gift cards +4; amount over 100 +3; unknown sender +2**. Each signal has its own `if`. The clean and all-signals cases are included, along with a test that reaches every reporting branch.


````python
# CODE_RUNNER: 3.6 Homework - Score a message then advise on it
def advise(message, amount_requested, known_contact):
    risk_score = 0
    # No else needed: if this signal is absent, its contribution stays zero.
    if "gift cards" in message.lower():
        risk_score += 4
    # No else needed: amounts at or below 100 contribute zero points.
    if amount_requested > 100:
        risk_score += 3
    # No else needed: a known contact contributes zero points.
    if not known_contact:
        risk_score += 2

    print("Risk score:", risk_score)
    # 3 thresholds -> 4 outcomes -> 4 answers
    if risk_score >= 9:
        print("Do not reply. This is almost certainly a scam.")
        branch = "stop"
    elif risk_score >= 6:
        print("This looks risky. Check with someone you trust before replying.")
        branch = "check"
    elif risk_score >= 3:
        print("Be a little careful, but this is probably fine.")
        branch = "careful"
    else:
        print("Nothing suspicious in this message.")
        branch = "clean"
    return risk_score, branch

# Starter case: all three independent signals are present.
message = "URGENT: send gift cards now to unlock your account"
amount_requested = 500
known_contact = False
assert advise(message, amount_requested, known_contact) == (9, "stop")

# Fresh inputs cover the clean extreme and all other reporting outcomes.
assert advise("Practice at 4 PM", 0, True) == (0, "clean")
assert advise("Please send payment", 101, True) == (3, "careful")
assert advise("Send gift cards", 0, False) == (6, "check")
assert advise("Practice at 4 PM", 100, True) == (0, "clean")
print("All four branches and the 100/101 amount boundary passed.")
````
