**MCQ: 4/4.** I developed, traced, and tested an inventory-priority algorithm.

```python
# CODE_RUNNER: Homework - inventory priority algorithm
def urgent_item(items):
    return max(items, key=lambda item: item['needed'] - item['available'])
items = [{'name':'Blankets','needed':20,'available':8},{'name':'Shampoo','needed':15,'available':5},{'name':'Snacks','needed':10,'available':12}]
for item in items: print(item['name'], item['needed'] - item['available'])
answer = urgent_item(items)
print('Most urgent:', answer['name'])
assert answer['name'] == 'Blankets'
```

## Evidence
The algorithm has a clear input, repeats one calculation for every item, and returns the item with the greatest shortage.
