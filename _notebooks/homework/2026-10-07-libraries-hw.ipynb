## Popcorn 1 — Library procedures
randint(a,b) returns an integer including both endpoints; ceil(x) rounds upward. choice selects a member, and sqrt returns a square root. I changed the jersey range to 1–55 and the court area to 225. Prediction: a roster member, jersey in that range, court side 15.0, and ceil(17.2)=18.


````python
# CODE_RUNNER: 2026-10-07-libraries-hw exercise 3
import random
import math
random.seed(14)
roster = ["Rohan", "Adhvay", "Ishan"]
player = random.choice(roster)
jersey = random.randint(1, 55)
court_side = math.sqrt(225)
points_target = math.ceil(17.2)
print(player, jersey, court_side, points_target)
assert player in roster and 1 <= jersey <= 55
assert court_side == 15 and points_target == 18
````

## Popcorn 2 — Flask routes
I customized score and roster and added /coach. Flask is third-party, unlike math/random. These tests call Flask's local test client: they test the route functions without starting a persistent server. The browser runner may not provide Flask; use the notebook's Python environment after installing requirements.


````python
# CODE_RUNNER: 2026-10-07-libraries-hw exercise 5
from flask import Flask, jsonify
app = Flask(__name__)
@app.route("/score")
def score():
    return jsonify(team="GameMaker", score=84)
@app.route("/roster")
def roster():
    return jsonify(players=["Rohan", "Adhvay", "Ishan"])
@app.route("/coach")
def coach():
    return jsonify(coach="Student-led", strategy="Practice and review")
client = app.test_client()
for path in ["/score", "/roster", "/coach"]:
    response = client.get(path)
    assert response.status_code == 200
    print(path, response.get_json())
assert client.get("/score").get_json()["score"] == 84
assert len(client.get("/roster").get_json()["players"]) == 3
assert client.get("/missing").status_code == 404
````

## Homework — Esports season report
This is a sample dataset, not a claim about real matches. json serializes the report, statistics computes its mean, and datetime computes a date interval. Prediction: total 84, average 21, best 30; Westview totals 30 across two games. A fixed report date makes the 17-day interval reproducible.


````python
# CODE_RUNNER: 2026-10-07-libraries-hw exercise 7
import json
import statistics
from datetime import date

def summarize_season(season):
    """Summarize a list of game dictionaries.

    Parameters:
        season: list of records containing opponent (str) and points (number).
    Returns:
        dict with total_points, average_points, best_game, points_by_opponent.
        An empty season has total/average 0, best_game None, and no opponents.
    """
    points = [game["points"] for game in season]
    by_opponent = {}
    for game in season:
        opponent = game["opponent"]
        by_opponent[opponent] = by_opponent.get(opponent, 0) + game["points"]
    return {
        "total_points": sum(points),
        "average_points": statistics.mean(points) if points else 0,
        "best_game": max(points) if points else None,
        "points_by_opponent": by_opponent,
    }

season = [
    {"opponent":"Westview", "points":18},
    {"opponent":"Rancho Bernardo", "points":24},
    {"opponent":"Westview", "points":12},
    {"opponent":"Mt. Carmel", "points":30},
]
summary = summarize_season(season)
report_date = date(2026, 10, 7)
next_game = date(2026, 10, 24)
days_until = (next_game - report_date).days
report = {"as_of":report_date.isoformat(), "next_game":next_game.isoformat(),
          "days_until_next_game":days_until, "summary":summary}
print(json.dumps(report, indent=2))
assert summary["total_points"] == 84
assert summary["average_points"] == 21
assert summary["best_game"] == 30
assert summary["points_by_opponent"]["Westview"] == 30
assert days_until == 17
assert summarize_season([]) == {"total_points":0, "average_points":0, "best_game":None, "points_by_opponent":{}}
assert summarize_season([{"opponent":"A", "points":7}])["average_points"] == 7
assert json.loads(json.dumps(report)) == report
print("Normal, empty, singleton, date, and JSON round-trip tests passed.")
````
