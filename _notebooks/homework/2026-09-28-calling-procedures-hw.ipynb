```python
# CODE_RUNNER: Procedures Popcorn #1
def report_status(location, status):
    print("Status update from " + location + ": " + status)

report_status("Poway", "Medical team dispatched")
report_status("Rancho Bernardo", "Power restored")
```

### Popcorn Hack 2 — Decomposition and a Helper Procedure

```python
# CODE_RUNNER: Popcorn Hack 2 - Decomposition & Helper Procedures
# Helper procedure: returns a Boolean based on the incident type.
def is_priority(incident_type):
    return incident_type == "Medical Emergency"

# Outer procedure: delegates the priority decision to the helper.
def process_report(location, incident_type):
    priority = is_priority(incident_type)
    if priority:
        print("PRIORITY: " + incident_type + " reported in " + location)
    else:
        print("STANDARD REPORT: " + incident_type + " reported in " + location)

process_report("Poway", "Medical Emergency")
process_report("Carmel Mountain", "Power Outage")

assert is_priority("Medical Emergency") is True
assert is_priority("Power Outage") is False
```

### Homework Hack — Emergency Dispatch Program

`check_priority()` focuses only on deciding whether an incident needs priority handling. `dispatch_response()` handles the larger task by calling that helper and printing the appropriate response. This decomposition keeps the decision reusable and makes the outer procedure easier to read.

```python
# CODE_RUNNER: Homework Hack - emergency response procedures
# Helper procedure: returns True when the incident requires priority handling.
def check_priority(incident_type):
    priority_incidents = ["Medical Emergency", "Structure Fire"]
    return incident_type in priority_incidents

# Outer procedure: calls the helper and dispatches the appropriate response.
def dispatch_response(location, incident_type):
    priority = check_priority(incident_type)
    if priority:
        print("PRIORITY RESPONSE: Send emergency crew to " + location + " for " + incident_type)
    else:
        print("STANDARD RESPONSE: Record " + incident_type + " at " + location)

# Each call supplies arguments for different locations and incident types.
dispatch_response("Poway", "Medical Emergency")
dispatch_response("Rancho Bernardo", "Power Outage")
dispatch_response("Mira Mesa", "Structure Fire")
dispatch_response("Carmel Valley", "Traffic Hazard")

assert check_priority("Medical Emergency") is True
assert check_priority("Structure Fire") is True
assert check_priority("Traffic Hazard") is False
```
