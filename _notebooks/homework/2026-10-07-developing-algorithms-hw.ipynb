## Popcorn 1 — Sequence and accumulation
Prediction: 14 hours and average 3.5. Initialize, visit each entry, add it, then calculate the average.


````python
# CODE_RUNNER: 2026-10-07-developing-algorithms-hw exercise 3
hours = [3, 5, 2, 4]
# Start an empty accumulator.
total = 0
# Add each volunteer's hours.
for value in hours:
    total += value
# Calculate after processing every value.
print(total)
print(total / len(hours))
assert total == 14
````

## Popcorn 2 — Capacity boundary
The original > comparison counts only 15 (one workshop). At capacity 12 a workshop is also full; >= correctly counts two.


````python
# CODE_RUNNER: 2026-10-07-developing-algorithms-hw exercise 5
signed_up = [12, 7, 15, 9]
capacity = 12
original = 0
for count in signed_up:
    if count > capacity:
        original += 1
full = 0
for count in signed_up:
    if count >= capacity:
        full += 1
print("Original:", original)
print("Corrected:", full)
assert (original, full) == (1, 2)
````

### Recorded execution result

```text

Corrected: 2
```

## Homework — Restock report
Prediction: Blankets need 12 and Shampoo need 10; total 22 and Blankets are most urgent. Equal stock/target is sufficient. The algorithm uses parallel lists, then accumulation, then a maximum scan.


````python
# CODE_RUNNER: 2026-10-07-developing-algorithms-hw exercise 10
def restock_report(items, in_stock, target):
    # Validate the parallel-list relationship.
    assert len(items) == len(in_stock) == len(target)
    # Collect only positive deficits in corresponding lists.
    short_items = []
    short_amounts = []
    for index in range(len(items)):
        if in_stock[index] < target[index]:
            short_items.append(items[index])
            short_amounts.append(target[index] - in_stock[index])
    # Guard the empty case before indexing.
    if len(short_items) == 0:
        print("Fully stocked!")
        return short_items, short_amounts, 0, None
    # Accumulate the number of units needed.
    total = 0
    for amount in short_amounts:
        total += amount
    # Track the position of the largest deficit; ties keep the first item.
    largest_index = 0
    for index in range(1, len(short_amounts)):
        if short_amounts[index] > short_amounts[largest_index]:
            largest_index = index
    # Combine the collected results into the required report.
    print("Restock needed:")
    for index in range(len(short_items)):
        print(f"{short_items[index]}: need {short_amounts[index]}")
    print(f"Total items needed: {total}")
    print(f"Most urgent: {short_items[largest_index]} (short by {short_amounts[largest_index]})")
    return short_items, short_amounts, total, short_items[largest_index]

items = ["Toothbrushes", "Blankets", "Socks", "Shampoo", "Notebooks"]
in_stock = [40, 8, 25, 5, 30]
target = [30, 20, 25, 15, 30]
assert restock_report(items, in_stock, target) == (["Blankets", "Shampoo"], [12, 10], 22, "Blankets")
print("--- Boundary tests ---")
assert restock_report(["Socks"], [25], [25]) == ([], [], 0, None)
assert restock_report([], [], []) == ([], [], 0, None)
assert restock_report(["A", "B"], [0, 0], [2, 2])[-1] == "A"
print("All boundary assertions passed.")
````
