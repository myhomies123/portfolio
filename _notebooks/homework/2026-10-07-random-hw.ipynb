## Popcorn 1 — Inclusive bounds
With four records, randint(1,4) may return 1,2,3,4 and may repeat. I changed the count to six; prediction is integers 1 through 6, not a guarantee that a small sample contains every value.


````python
# CODE_RUNNER: 2026-10-07-random-hw exercise 3
import random
random.seed(15)
record_count = 6
samples = []
for trial in range(20):
    samples.append(random.randint(1, record_count))
print(samples)
assert all(1 <= value <= record_count for value in samples)
````

## Popcorn 3 — Random actions with selection
GET searches, POST creates, and PUT updates. Randomly select both a part and action; a fixed loop afterward demonstrates every action branch rather than relying on chance.


````python
# CODE_RUNNER: 2026-10-07-random-hw exercise 7
import random
random.seed(17)
parts = [{"product":"Flywheel","spec":"1.1"},{"product":"Clutch","spec":"1.2"},{"product":"Racing Flywheel","spec":"2.1"}]
actions = ["GET", "POST", "PUT"]
def describe(part, action):
    if action == "GET":
        return f"Search for {part['product']} ({part['spec']})"
    elif action == "POST":
        return f"Create {part['product']} ({part['spec']})"
    elif action == "PUT":
        return f"Update {part['product']} ({part['spec']})"
    else:
        raise ValueError("Unsupported action")
print(describe(random.choice(parts), random.choice(actions)))
for action in actions:
    print(describe(parts[0], action))
````

### Recorded execution result

```text

Search for Flywheel (1.1)
Create Flywheel (1.1)
Update Flywheel (1.1)
```

## Homework — Randomized API simulator
This simulates actions; it does not claim to send requests to a real API. Five trials choose a complete part and one of four actions. The state starts with specs 1.1 and 2.1. POST rejects duplicates, GET/PUT require an existing spec, and DELETE removes an existing spec. Predictions depend on previous state, so the seed makes this report repeatable.


````python
# CODE_RUNNER: 2026-10-07-random-hw exercise 10
import random
def apply_action(existing, part, action):
    spec = part["spec"]
    if action == "GET":
        return "Found" if spec in existing else "Not found"
    elif action == "POST":
        if spec in existing:
            return "Rejected duplicate"
        existing.append(spec)
        return "Created"
    elif action == "PUT":
        return "Updated" if spec in existing else "Cannot update missing spec"
    elif action == "DELETE":
        if spec in existing:
            existing.remove(spec)
            return "Removed"
        return "Cannot remove missing spec"
    raise ValueError("Unsupported action")

parts = [
 {"product":"Replacement Flywheels and Clutch Assemblies","category":"Auto Racing","spec":"1.1"},
 {"product":"Multiple Disc Clutch Assemblies","category":"Drag Racing","spec":"1.2"},
 {"product":"Racing Flywheel Record","category":"Auto Racing","spec":"2.1"}]
actions = ["GET", "POST", "PUT", "DELETE"]
existing = ["1.1", "2.1"]
rng = random.Random(24)
test_count = 5
for trial in range(test_count):
    part = rng.choice(parts)
    action = rng.choice(actions)
    result = apply_action(existing, part, action)
    print(f"{trial+1}: {action} | {part['product']} | {part['category']} | {part['spec']} | {result}")
print("Final specs:", existing)
# Controlled tests cover every branch even if random trials miss one.
test_state = ["1.1", "2.1"]
present = parts[0]
new = parts[1]
assert apply_action(test_state, present, "GET") == "Found"
assert apply_action(test_state, new, "GET") == "Not found"
assert apply_action(test_state, present, "POST") == "Rejected duplicate"
assert apply_action(test_state, new, "PUT") == "Cannot update missing spec"
assert apply_action(test_state, new, "DELETE") == "Cannot remove missing spec"
assert apply_action(test_state, new, "POST") == "Created"
assert apply_action(test_state, new, "PUT") == "Updated"
assert apply_action(test_state, new, "DELETE") == "Removed"
assert test_state == ["1.1", "2.1"]
print("All action and duplicate-state assertions passed.")
````

4: POST | Replacement Flywheels and Clutch Assemblies | Auto Racing | 1.1 | Rejected duplicate
5: GET | Racing Flywheel Record | Auto Racing | 2.1 | Found
