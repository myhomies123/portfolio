## Popcorn 1: Base class

The resource link uses `ocs__btn`. An anchor navigates; a button performs an action.

```python
%%html
<!-- UI_RUNNER: Buttons Popcorn 1 -->
<div id="sass-buttons-popcorn1" class="ui-runner-container" data-sass-homework>
  <div class="control-panel">
    <button type="button" class="ocs__btn accent fill" data-hook="run" aria-label="Run Buttons Popcorn 1">Run</button>
    <button type="button" class="ocs__btn outline" data-hook="reset" aria-label="Reset Buttons Popcorn 1">Reset</button>
    <output data-run-status aria-live="polite">Notebook HTML output</output>
  </div>
  <details>
    <summary>Edit completed code</summary>
    <div class="editor-container">
      <textarea class="editor-textarea" aria-label="Completed code for Buttons Popcorn 1" rows="16">outputElement.innerHTML = &quot;&lt;a class=\&quot;ocs__btn\&quot; href=\&quot;https://www.mygoodbrain.org/resources\&quot; target=\&quot;_blank\&quot; rel=\&quot;noopener\&quot;&gt;Resource Directory&lt;/a&gt;&quot;;
</textarea>
    </div>
  </details>
  <div class="output-container">
    <p>UI Output</p>
    <div class="ui-output"><a class="ocs__btn" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">Resource Directory</a></div>
  </div>
</div>
<script>
const root = document.getElementById('sass-buttons-popcorn1');
const area = root.querySelector('textarea');
const initialCode = area.value;
const outputElement = root.querySelector('.ui-output');
const status = root.querySelector('[data-run-status]');
let editor = null;
root.querySelector('details').addEventListener('toggle', function () {
  if (this.open && !editor && typeof CodeMirror !== 'undefined') {
    editor = CodeMirror.fromTextArea(area, {lineNumbers: true, mode: 'javascript', theme: 'darcula'});
    editor.setSize(null, 260);
  }
  if (editor) editor.refresh();
});
function run() {
  try {
    new Function('outputElement', editor ? editor.getValue() : area.value)(outputElement);
    status.textContent = 'Executed successfully';
  } catch (error) {
    status.textContent = 'Execution error: ' + error.message;
  }
}
root.querySelector('[data-hook="run"]').addEventListener('click', run);
root.querySelector('[data-hook="reset"]').addEventListener('click', function () {
  area.value = initialCode;
  if (editor) editor.setValue(initialCode);
  run();
});
run();
</script>
```

## Popcorn 2: Modifiers and grouping

The check-in button adds `pill fill`. Both actions are grouped inside `ocs__links`.

```python
%%html
<!-- UI_RUNNER: Buttons Popcorn 2 -->
<div id="sass-buttons-popcorn2" class="ui-runner-container" data-sass-homework>
  <div class="control-panel">
    <button type="button" class="ocs__btn accent fill" data-hook="run" aria-label="Run Buttons Popcorn 2">Run</button>
    <button type="button" class="ocs__btn outline" data-hook="reset" aria-label="Reset Buttons Popcorn 2">Reset</button>
    <output data-run-status aria-live="polite">Notebook HTML output</output>
  </div>
  <details>
    <summary>Edit completed code</summary>
    <div class="editor-container">
      <textarea class="editor-textarea" aria-label="Completed code for Buttons Popcorn 2" rows="16">outputElement.innerHTML = &quot;&lt;div class=\&quot;ocs__links\&quot;&gt;\n  &lt;button type=\&quot;button\&quot; class=\&quot;ocs__btn pill fill\&quot;&gt;Complete Daily Check-In&lt;/button&gt;\n  &lt;a class=\&quot;ocs__btn outline\&quot; href=\&quot;https://www.mygoodbrain.org/resources\&quot; target=\&quot;_blank\&quot; rel=\&quot;noopener\&quot;&gt;Explore Resources&lt;/a&gt;\n&lt;/div&gt;&quot;;
const action = outputElement.querySelector(&#x27;button&#x27;);
if (action) action.addEventListener(&#x27;click&#x27;, function () {
  let result = outputElement.querySelector(&#x27;output&#x27;);
  if (!result) {
    result = document.createElement(&#x27;output&#x27;);
    result.setAttribute(&#x27;aria-live&#x27;, &#x27;polite&#x27;);
    outputElement.appendChild(result);
  }
  result.textContent = &#x27;Demo action selected: &#x27; + action.textContent + &#x27;. No personal data is collected.&#x27;;
});</textarea>
    </div>
  </details>
  <div class="output-container">
    <p>UI Output</p>
    <div class="ui-output"><div class="ocs__links">
  <button type="button" class="ocs__btn pill fill">Complete Daily Check-In</button>
  <a class="ocs__btn outline" href="https://www.mygoodbrain.org/resources" target="_blank" rel="noopener">Explore Resources</a>
</div></div>
  </div>
</div>
<script>
const root = document.getElementById('sass-buttons-popcorn2');
const area = root.querySelector('textarea');
const initialCode = area.value;
const outputElement = root.querySelector('.ui-output');
const status = root.querySelector('[data-run-status]');
let editor = null;
root.querySelector('details').addEventListener('toggle', function () {
  if (this.open && !editor && typeof CodeMirror !== 'undefined') {
    editor = CodeMirror.fromTextArea(area, {lineNumbers: true, mode: 'javascript', theme: 'darcula'});
    editor.setSize(null, 260);
  }
  if (editor) editor.refresh();
});
function run() {
  try {
    new Function('outputElement', editor ? editor.getValue() : area.value)(outputElement);
    status.textContent = 'Executed successfully';
  } catch (error) {
    status.textContent = 'Execution error: ' + error.message;
  }
}
root.querySelector('[data-hook="run"]').addEventListener('click', run);
root.querySelector('[data-hook="reset"]').addEventListener('click', function () {
  area.value = initialCode;
  if (editor) editor.setValue(initialCode);
  run();
});
run();
</script>
```
