### Popcorn Hack 1 — Predict, Run, and Fix the Crash

### Popcorn Hack 2 — Accumulator Loop

```python
# CODE_RUNNER: Lists Popcorn 2 - sum and average with a loop
numbers = [7, 6, 8, 5, 7, 9]

total = 0
for number in numbers:
    total += number

average = round(total / len(numbers), 2)
print("Total:", total)
print("Average:", average)

assert total == 42
assert average == 7.0
```

### Popcorn Hack 3 — Count Then Filter

```python
# CODE_RUNNER: Lists Popcorn 3 - count then filter
seasons = ["winter", "summer", "fall", "summer", "spring", "summer", "fall"]

frequency = {}
for season in seasons:
    if season in frequency:
        frequency[season] += 1
    else:
        frequency[season] = 1

popular = []
for season in frequency:
    if frequency[season] >= 2:
        popular.append(season)

print(frequency)
print(popular)

assert popular == ["summer", "fall"]
```

### Homework — Snack Vote Analysis

The program uses one snack-vote dataset to demonstrate indexing, an accumulator, a frequency dictionary, and filtering without changing the original vote list.

```python
# CODE_RUNNER: Lists Homework - indexing, accumulator, count and filter on one dataset
votes = ["chips", "cookies", "chips", "fruit", "cookies", "chips",
         "pretzels", "cookies", "chips", "fruit", "chips", "cookies"]
prices = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]

# Lesson 1: dynamic indexing
vote_count = len(votes)
last_vote = votes[-1]

# Lesson 2: accumulator loop (no sum function)
total = 0
for price in prices:
    total += price
average = round(total / len(prices), 2)

# Lesson 3: frequency dictionary using the if/else key pattern
frequency = {}
for vote in votes:
    if vote in frequency:
        frequency[vote] += 1
    else:
        frequency[vote] = 1

# Build a new filtered list; leave votes unchanged.
popular_snacks = []
for snack in frequency:
    if frequency[snack] >= 3:
        popular_snacks.append(snack)

print("Votes:", vote_count)
print("Last vote:", last_vote)
print("Total:", int(total))
print("Average:", int(average))
print("Frequency:", frequency)
print("Popular snacks:", popular_snacks)

assert vote_count == 12 and last_vote == "cookies"
assert total == 31 and average == 2.58
assert frequency == {"chips": 5, "cookies": 4, "fruit": 2, "pretzels": 1}
assert popular_snacks == ["chips", "cookies"]
```
