`::before` is a pseudo-element, rather than a class. SASS `&::before` nests it under the track selector.

## Popcorn: Find the bug

The decorative span was missing `ocs__toggle-track`. I restored that class and `aria-hidden="true"`. The input remains the accessible control; the track supplies its visible appearance.

```python
%%html
<!-- UI_RUNNER: Toggles Bug Fix -->
<div id="sass-toggles-bug" class="ui-runner-container" data-sass-homework>
  <div class="control-panel">
    <button type="button" class="ocs__btn accent fill" data-hook="run" aria-label="Run Toggles Bug Fix">Run</button>
    <button type="button" class="ocs__btn outline" data-hook="reset" aria-label="Reset Toggles Bug Fix">Reset</button>
    <output data-run-status aria-live="polite">Notebook HTML output</output>
  </div>
  <details>
    <summary>Edit completed code</summary>
    <div class="editor-container">
      <textarea class="editor-textarea" aria-label="Completed code for Toggles Bug Fix" rows="16">outputElement.innerHTML = [
  &quot;&lt;label class=\&quot;ocs__toggle ocs__toggle--green\&quot;&gt;&quot;,
  &quot;  &lt;input id=\&quot;toggle-bug-fixed\&quot; class=\&quot;ocs__toggle-input\&quot; type=\&quot;checkbox\&quot;&gt;&quot;,
  &quot;  &lt;span class=\&quot;ocs__toggle-track\&quot; aria-hidden=\&quot;true\&quot;&gt;&lt;/span&gt;&quot;,
  &quot;  &lt;span class=\&quot;ocs__toggle-label\&quot;&gt;Enable Notifications&lt;/span&gt;&quot;,
  &quot;&lt;/label&gt;&quot;,
  &quot;&lt;p&gt;&lt;output aria-live=\&quot;polite\&quot;&gt;Checking the repaired markup…&lt;/output&gt;&lt;/p&gt;&quot;
].join(&#x27;\n&#x27;);

// The missing track class restores the visible switch; the checkbox stores state.
const track = outputElement.querySelector(&#x27;.ocs__toggle-track&#x27;);
const result = outputElement.querySelector(&#x27;output&#x27;);
result.textContent = track ? &#x27;Toggle works! 🎉&#x27; : &#x27;The track class is missing.&#x27;;
const input = outputElement.querySelector(&#x27;input&#x27;);
input.addEventListener(&#x27;change&#x27;, function () {
  result.textContent = &#x27;Toggle works! 🎉 Notifications demo: &#x27; + (input.checked ? &#x27;ON&#x27; : &#x27;OFF&#x27;);
});</textarea>
    </div>
  </details>
  <div class="output-container">
    <p>UI Output</p>
    <div class="ui-output"><label class="ocs__toggle ocs__toggle--green">
  <input id="toggle-bug-fixed" class="ocs__toggle-input" type="checkbox">
  <span class="ocs__toggle-track" aria-hidden="true"></span>
  <span class="ocs__toggle-label">Enable Notifications</span>
</label>
<p><output aria-live="polite">Checking the repaired markup…</output></p></div>
  </div>
</div>
<script>
const root = document.getElementById('sass-toggles-bug');
const area = root.querySelector('textarea');
const initialCode = area.value;
const outputElement = root.querySelector('.ui-output');
const status = root.querySelector('[data-run-status]');
let editor = null;
root.querySelector('details').addEventListener('toggle', function () {
  if (this.open && !editor && typeof CodeMirror !== 'undefined') {
    editor = CodeMirror.fromTextArea(area, {lineNumbers: true, mode: 'javascript', theme: 'darcula'});
    editor.setSize(null, 260);
  }
  if (editor) editor.refresh();
});
function run() {
  try {
    new Function('outputElement', editor ? editor.getValue() : area.value)(outputElement);
    status.textContent = 'Executed successfully';
  } catch (error) {
    status.textContent = 'Execution error: ' + error.message;
  }
}
root.querySelector('[data-hook="run"]').addEventListener('click', run);
root.querySelector('[data-hook="reset"]').addEventListener('click', function () {
  area.value = initialCode;
  if (editor) editor.setValue(initialCode);
  run();
});
run();
</script>
```

## Reflection

SASS styles the switch, while JavaScript decides what the switch does. The `:checked + .ocs__toggle-track` selector changes its appearance. JavaScript reads `.checked`, uses Boolean selection to update the feature, and recomputes the active count after each `change` event.

The native checkbox supports keyboard input. Labels name each control, `aria-controls` identifies the connected content, and polite live outputs announce the count and feedback.
