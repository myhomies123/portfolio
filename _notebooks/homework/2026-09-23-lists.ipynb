## Popcorn 2 — Accumulator
Prediction: total 42, average 7.0. Each loop adds one volunteer's hours; divide only after the loop.


````python
# CODE_RUNNER: 2026-09-23-lists exercise 5
volunteer_hours = [7, 6, 8, 5, 7, 9]
total = 0
for hours in volunteer_hours:
    total += hours
print(total)
print(total / len(volunteer_hours))
assert total == 42
````

## Popcorn 3 — Frequency and filtering
Prediction: socks occurs 3 times and meals twice, so both meet the threshold of 2. A dictionary stores one counter per item.


````python
# CODE_RUNNER: 2026-09-23-lists exercise 7
items_requested = ["blankets", "socks", "meals", "socks", "water", "socks", "meals"]
frequency = {}
for item in items_requested:
    if item in frequency:
        frequency[item] += 1
    else:
        frequency[item] = 1
popular = []
for item in frequency:
    if frequency[item] >= 2:
        popular.append(item)
print(frequency)
print(popular)
assert popular == ["socks", "meals"]
````

## Homework — Donation analysis
Prediction: 12 donations, last item socks, total cost 31, integer average 2. Blankets and socks qualify at threshold 3. The original lists remain unchanged. No sum() or list comprehensions are used.


````python
# CODE_RUNNER: 2026-09-23-lists exercise 9
donations_by_item = ["blankets","socks","blankets","water","socks","blankets","meals","socks","blankets","water","blankets","socks"]
costs = [2,3,2,4,3,2,1,3,2,4,2,3]
original_items = donations_by_item[:]
original_costs = costs[:]
# Index using the current list length, not a hard-coded last position.
print(len(donations_by_item))
print(donations_by_item[-1])
# Accumulate every cost once.
total = 0
for cost in costs:
    total += cost
average = int(total / len(costs))
print("Total:", total)
print("Average:", average)
# Count occurrences without altering the donation data.
frequency = {}
for item in donations_by_item:
    if item in frequency:
        frequency[item] += 1
    else:
        frequency[item] = 1
# Build a new result list with the inclusive popularity threshold.
popular = []
for item in frequency:
    if frequency[item] >= 3:
        popular.append(item)
print(frequency)
print(popular)
assert total == 31 and average == 2
assert frequency == {"blankets":5, "socks":4, "water":2, "meals":1}
assert popular == ["blankets", "socks"]
assert donations_by_item == original_items and costs == original_costs
````
