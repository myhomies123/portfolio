## MCQ result and reasoning

Completed on the original lesson: **MCQ Grids: 4/4 &#124; answers: A,B,C,D**.

| Question topic | Answer | Why |
| --- | --- | --- |
| Four standard columns | A: `cols-4` | The modifier sets four equal columns. |
| `2fr 1fr` | B: two columns, first twice as wide | Two tracks divide the space into three shares. |
| Full-row title | C: `ocs__grid-cell--header` | A header spans every column; `--wide` spans two. |
| Grammar violation | D: custom row plus inline flex | Shared OCS grammar supplies the layout. |
