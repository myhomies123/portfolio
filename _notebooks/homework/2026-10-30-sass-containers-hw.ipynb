## Popcorn: Parade invite rescue

I replaced the page wrapper, purple inline heading, and flex boxes with one container, one card, and three standard-grid cells. The start time is accented because it is the key planning detail.

```python
%%html
<!-- UI_RUNNER: Containers Popcorn -->
<div id="sass-containers-popcorn" class="ui-runner-container" data-sass-homework>
  <div class="control-panel">
    <button type="button" class="ocs__btn accent fill" data-hook="run" aria-label="Run Containers Popcorn">Run</button>
    <button type="button" class="ocs__btn outline" data-hook="reset" aria-label="Reset Containers Popcorn">Reset</button>
    <output data-run-status aria-live="polite">Notebook HTML output</output>
  </div>
  <details>
    <summary>Edit completed code</summary>
    <div class="editor-container">
      <textarea class="editor-textarea" aria-label="Completed code for Containers Popcorn" rows="16">outputElement.innerHTML = &quot;&lt;div class=\&quot;ocs__container\&quot;&gt;\n  &lt;div class=\&quot;ocs__card\&quot;&gt;\n    &lt;h2 class=\&quot;ocs__section-title\&quot;&gt;Poway Rotary Parade Invite&lt;/h2&gt;\n    &lt;p class=\&quot;ocs__description\&quot;&gt;Join neighbors for a morning of community service and parade celebrations.&lt;/p&gt;\n    &lt;div class=\&quot;ocs__grid ocs__grid--standard cols-3\&quot;&gt;\n      &lt;div class=\&quot;ocs__grid-cell\&quot;&gt;Location: Poway Road&lt;/div&gt;\n      &lt;div class=\&quot;ocs__grid-cell ocs__grid-cell--accent\&quot;&gt;Start time: 9:00 AM&lt;/div&gt;\n      &lt;div class=\&quot;ocs__grid-cell\&quot;&gt;What you get: community, music, and volunteer opportunities&lt;/div&gt;\n    &lt;/div&gt;\n  &lt;/div&gt;\n&lt;/div&gt;&quot;;
</textarea>
    </div>
  </details>
  <div class="output-container">
    <p>UI Output</p>
    <div class="ui-output"><div class="ocs__container">
  <div class="ocs__card">
    <h2 class="ocs__section-title">Poway Rotary Parade Invite</h2>
    <p class="ocs__description">Join neighbors for a morning of community service and parade celebrations.</p>
    <div class="ocs__grid ocs__grid--standard cols-3">
      <div class="ocs__grid-cell">Location: Poway Road</div>
      <div class="ocs__grid-cell ocs__grid-cell--accent">Start time: 9:00 AM</div>
      <div class="ocs__grid-cell">What you get: community, music, and volunteer opportunities</div>
    </div>
  </div>
</div></div>
  </div>
</div>
<script>
const root = document.getElementById('sass-containers-popcorn');
const area = root.querySelector('textarea');
const initialCode = area.value;
const outputElement = root.querySelector('.ui-output');
const status = root.querySelector('[data-run-status]');
let editor = null;
root.querySelector('details').addEventListener('toggle', function () {
  if (this.open && !editor && typeof CodeMirror !== 'undefined') {
    editor = CodeMirror.fromTextArea(area, {lineNumbers: true, mode: 'javascript', theme: 'darcula'});
    editor.setSize(null, 260);
  }
  if (editor) editor.refresh();
});
function run() {
  try {
    new Function('outputElement', editor ? editor.getValue() : area.value)(outputElement);
    status.textContent = 'Executed successfully';
  } catch (error) {
    status.textContent = 'Execution error: ' + error.message;
  }
}
root.querySelector('[data-hook="run"]').addEventListener('click', run);
root.querySelector('[data-hook="reset"]').addEventListener('click', function () {
  area.value = initialCode;
  if (editor) editor.setValue(initialCode);
  run();
});
run();
</script>
```
