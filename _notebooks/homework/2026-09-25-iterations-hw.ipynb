## Python 3.08 Iterations

This notebook completes the wildfire dispatch Popcorn Hack and a cybersecurity-themed monitoring checker. Both cells use a `for` loop, thresholds, counters, `continue`, and `break`.

### Popcorn Hack — Wildfire Dispatch Scanner

````python
# CODE_RUNNER: Popcorn Hack - wildfire dispatch scanner
# Theme: The SFSRC team scans wildfire alert levels from different zones.
alerts = [1, 4, 2, 6, 3, 9, 5]
safe_cutoff = 3
critical_cutoff = 9

action_count = 0
safe_count = 0
total_action_score = 0
critical_found = False

for alert in alerts:
    if alert < safe_cutoff:
        safe_count += 1
        print("Alert " + str(alert) + ": safe, skipping")
        continue

    action_count += 1
    total_action_score += alert
    print("Alert " + str(alert) + ": action needed")

    if alert >= critical_cutoff:
        critical_found = True
        print("Critical alert found - stopping scan")
        break

print("Safe alerts skipped: " + str(safe_count))
print("Action alerts: " + str(action_count))
print("Total action score: " + str(total_action_score))
print("Critical found: " + str(critical_found))

assert (safe_count, action_count, total_action_score, critical_found) == (2, 4, 22, True)
````

### Homework Hack — Server Risk Monitoring

The system scans server risk scores. Scores below 40 are safe and skipped. Scores of 90 or higher are critical, so the scan immediately stops for incident response.

````python
# CODE_RUNNER: Homework Hack - cybersecurity server risk checker
# The server risk scores represent security-monitoring results from different systems.
server_risk_scores = [18, 42, 55, 27, 71, 88, 63, 95, 34]
safe_cutoff = 40
critical_cutoff = 90

checked_count = 0
action_count = 0
skipped_count = 0
total_risk_score = 0
critical_found = False

for risk_score in server_risk_scores:
    checked_count += 1

    if risk_score < safe_cutoff:
        skipped_count += 1
        print("Risk score " + str(risk_score) + ": safe, no action")
        continue

    action_count += 1
    total_risk_score += risk_score
    print("Risk score " + str(risk_score) + ": security action needed")

    if risk_score >= critical_cutoff:
        critical_found = True
        print("Critical risk found. Stop checking and begin incident response.")
        break

print("Final Server Risk Report")
print("Checked: " + str(checked_count))
print("Skipped: " + str(skipped_count))
print("Actions: " + str(action_count))
print("Total risk score: " + str(total_risk_score))
print("Critical found: " + str(critical_found))

assert len(server_risk_scores) == 9
assert (checked_count, skipped_count, action_count, total_risk_score, critical_found) == (8, 2, 6, 414, True)
````

````python
# CODE_RUNNER: 2026-09-25-iterations-hw exercise 9
def scan(values, safe, critical):
    checked = skipped = actions = total = 0
    found = False
    for value in values:
        checked += 1
        if value < safe:
            skipped += 1
            continue
        actions += 1
        total += value
        if value >= critical:
            found = True
            break
    return checked, skipped, actions, total, found
assert scan([],40,90) == (0,0,0,0,False)
assert scan([1,2,3],40,90) == (3,3,0,0,False)
assert scan([40,89],40,90) == (2,0,2,129,False)
assert scan([90,100],40,90) == (1,0,1,90,True)
print("Empty, all-safe, no-critical, and exact-threshold tests passed.")
````
