## Popcorn Hacks and MCQ
I used an outer decision for whether a student has time and an inner decision for the best activity. **MCQ: 4/4.**

```python
# CODE_RUNNER: Nested Conditionals Popcorn - student-life decision
def after_school_plan(has_time, homework_done, energy):
    if has_time:
        if not homework_done: return 'Finish homework first.'
        if energy >= 6: return 'Go to club practice.'
        return 'Take a short break.'
    return 'Plan for tomorrow.'

for case in [(True, False, 8), (True, True, 7), (False, True, 8)]: print(case, after_school_plan(*case))
assert after_school_plan(True, True, 7) == 'Go to club practice.'
```

## Homework Hack — Balanced Student Schedule

This function uses nested decisions to give one understandable recommendation from schoolwork, available time, and energy.

```python
# CODE_RUNNER: Homework - student schedule helper
def schedule_helper(minutes_free, assignments_left, energy):
    if minutes_free >= 30:
        if assignments_left > 0:
            return 'Work for 25 minutes, then take a break.'
        if energy >= 5:
            return 'Choose a club, sport, or creative activity.'
        return 'Rest and recharge.'
    if assignments_left > 0: return 'Write a short plan for your next work session.'
    return 'Enjoy your free time.'

assert schedule_helper(45, 2, 8).startswith('Work')
assert schedule_helper(45, 0, 3) == 'Rest and recharge.'
assert schedule_helper(10, 1, 8).startswith('Write')
print('All nested-condition tests passed.')
```

## Evidence
The outer condition prevents unnecessary inner checks, and the tests cover work, activity, rest, and limited-time outcomes.
