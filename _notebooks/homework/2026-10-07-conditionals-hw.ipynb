## Popcorn Hacks and MCQ
I fixed the silent `if` by adding an `else`, then ordered the risk thresholds from highest to lowest so every branch is reachable. **MCQ: 4/4.**

```python
# CODE_RUNNER: Conditionals Popcorn - complete outcomes and reachable thresholds
def money_message(amount):
    if amount > 100: return 'Large amount: ask for help.'
    return 'No large amount requested.'

def risk_message(score):
    if score >= 9: return 'Do not reply.'
    if score >= 6: return 'Check with someone you trust.'
    if score >= 3: return 'Be careful.'
    return 'Nothing suspicious.'

print(money_message(0))
print(money_message(150))
print(risk_message(9))
assert risk_message(9) == 'Do not reply.'
```

## Homework Hack — Scam-message trainer

The program combines multiple conditions into one clear recommendation and tests low, medium, and high-risk messages.

```python
# CODE_RUNNER: Homework - Scam-message trainer
def classify_message(amount_requested, unknown_sender, asks_for_gift_cards):
    risk = (4 if amount_requested > 100 else 0) + (3 if unknown_sender else 0) + (2 if asks_for_gift_cards else 0)
    return risk, risk_message(risk)

tests = [(0, False, False), (50, True, False), (200, True, True)]
for test in tests:
    print(test, '=>', classify_message(*test))
assert classify_message(200, True, True) == (9, 'Do not reply.')
assert classify_message(0, False, False) == (0, 'Nothing suspicious.')
```

## Evidence
Every input receives an output, and the tests demonstrate that high-risk conditions are checked first instead of being hidden behind a lower threshold.
