# 3.17 Algorithmic Efficiency Homework

**MCQ: 4/4.** I compared the work required by linear and binary search as a catalog grows.

```python
# CODE_RUNNER: Homework - compare search efficiency
import math
def search_work(n):
    linear = n
    binary = 0 if n == 0 else math.floor(math.log2(n)) + 1
    return linear, binary
for size in [8, 16, 32]:
    linear, binary = search_work(size); print(size, 'linear:', linear, 'binary:', binary)
assert search_work(16) == (16, 5)
assert search_work(0) == (0, 0)
```

## Evidence
When the catalog doubles, linear work doubles while the binary-search estimate grows by one; binary search requires sorted data.
