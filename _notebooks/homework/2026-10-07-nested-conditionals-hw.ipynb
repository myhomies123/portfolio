### Popcorn Hack 1 — After-school homework plan (0.15)

Prediction for the starter: Finish homework. The outer weekday branch is True; the inner finished branch is False. Tests below cover all three outcomes.


````python
# CODE_RUNNER: Popcorn Hack 1 - After-school homework plan
def homework_plan(is_weekday, homework_finished):
    if is_weekday:
        if homework_finished:
            result = "Free time"
        else:
            result = "Finish homework"
    else:
        result = "Weekend plan"
    print(is_weekday, homework_finished, "=>", result)
    return result
assert homework_plan(True, False) == "Finish homework"
assert homework_plan(True, True) == "Free time"
assert homework_plan(False, False) == "Weekend plan"
assert homework_plan(False, True) == "Weekend plan"
````



### Popcorn Hack 2 — Repair the account check (0.15)

The original permission `if` was at the same indentation as the login `if`, so it ran even for logged-out users. Nesting it under login and adding an outer else ensures permission is checked only after account access.


````python
# CODE_RUNNER: Popcorn Hack 2 - Repair the student account check
def account_check(logged_in, has_permission):
    if logged_in:
        print("Account found")
        if has_permission:
            result = "Page opened"
        else:
            result = "Permission needed"
    else:
        result = "Log in first"
    print(result)
    return result
assert account_check(True, True) == "Page opened"
assert account_check(True, False) == "Permission needed"
assert account_check(False, True) == "Log in first"
assert account_check(False, False) == "Log in first"
````

### Recorded execution result

```text
Log in first
Log in first
```


### Popcorn Hack 3 — Lunch and sports practice (0.20)

The outer lunch decision runs first; its selected inner branch checks practice. Prediction: the starter prints packed lunch plus an extra snack for practice. All four combinations are tested.


````python
# CODE_RUNNER: Popcorn Hack 3 - Lunch and sports-practice plan
def lunch_plan(lunch_packed, practice_after_school):
    if lunch_packed:
        if practice_after_school:
            result = "Bring packed lunch and an extra practice snack."
        else:
            result = "Bring packed lunch; no practice snack needed."
    else:
        if practice_after_school:
            result = "Buy lunch and bring an extra practice snack."
        else:
            result = "Buy lunch; no practice snack needed."
    print(result)
    return result
results = []
for lunch_packed in [True, False]:
    for practice_after_school in [True, False]:
        results.append(lunch_plan(lunch_packed, practice_after_school))
assert len(set(results)) == 4
````

### Recorded execution result

```text
Bring packed lunch and an extra practice snack.
Bring packed lunch; no practice snack needed.
Buy lunch and bring an extra practice snack.
Buy lunch; no practice snack needed.
```

### Homework Hack — High school daily routine dashboard (0.50)

The starter's four values are retained. Login is the outer guard. Inside it, homework, meals, practice, and dinner produce clear recommendations. All 16 Boolean combinations are exercised below.


````python
# CODE_RUNNER: Homework Hack - High school daily routine dashboard
student = {
    "logged_in": True,
    "homework_finished": False,
    "lunch_packed": True,
    "practice_today": True,
}

def daily_dashboard(student):
    recommendations = []
    if student["logged_in"]:
        recommendations.append("Account opened.")
        if student["homework_finished"]:
            recommendations.append("Homework done: enjoy free time.")
        else:
            recommendations.append("Finish homework before free time.")
        if student["lunch_packed"]:
            if student["practice_today"]:
                recommendations.append("Packed lunch: add a practice snack.")
            else:
                recommendations.append("Packed lunch: eat it at lunch break.")
        else:
            if student["practice_today"]:
                recommendations.append("Buy lunch and bring a practice snack.")
            else:
                recommendations.append("Buy lunch at school.")
        if student["practice_today"]:
            recommendations.append("Attend practice, then have dinner.")
        else:
            recommendations.append("No practice: have dinner at home.")
    else:
        recommendations.append("Log in first to view your daily schedule.")
    for recommendation in recommendations:
        print(recommendation)
    return recommendations

print("Starter dashboard:")
starter_result = daily_dashboard(student)
assert len(starter_result) == 4
print("Exhaustive outcome tests:")
for logged_in in [False, True]:
    for homework_finished in [False, True]:
        for lunch_packed in [False, True]:
            for practice_today in [False, True]:
                case = dict(logged_in=logged_in, homework_finished=homework_finished,
                            lunch_packed=lunch_packed, practice_today=practice_today)
                print("Case:", case)
                result = daily_dashboard(case)
                if logged_in:
                    assert len(result) == 4
                    assert ("Homework done: enjoy free time." in result) == homework_finished
                    assert ("Attend practice, then have dinner." in result) == practice_today
                else:
                    assert result == ["Log in first to view your daily schedule."]
print("All 16 input combinations passed.")
````
