# 3.01 Variables and Assignments Homework

This notebook shows how variables store values at the moment of assignment and how a program must recalculate a total after one of its values changes.

## Popcorn Hack — Keep the checkpoint

```python
# CODE_RUNNER: UESL 3.1 Popcorn - Keep the checkpoint
stars = 8
checkpoint_stars = stars
stars = stars + 4

print(stars)
print(checkpoint_stars)

assert (stars, checkpoint_stars) == (12, 8)
print("Validation: the checkpoint kept its original value.")
```

Prediction: `stars` becomes 12 while `checkpoint_stars` stays 8. Assignment stores the value that existed at that moment; changing `stars` later does not update the checkpoint automatically.

## Homework — Star counter with a bonus

The game prototype stores individual level totals, demonstrates a stale stored total, and then recalculates it after a bonus.

```python
# CODE_RUNNER: UESL 3.1 Homework - Star counter with a bonus
game_title = "UESL Star Trail"
reduced_motion = True

level_one_stars = 12
level_two_stars = 18
level_three_stars = 10

total_stars = level_one_stars + level_two_stars + level_three_stars
print(game_title)
print("Reduced motion:", reduced_motion)
print("Original total:", total_stars)

level_two_stars = level_two_stars + 5
print("Stored total after bonus:", total_stars)

total_stars = level_one_stars + level_two_stars + level_three_stars
print("Recalculated total after bonus:", total_stars)

assert total_stars == 45
print("Validation: the recalculated total includes the bonus.")
```

## Test — bonus of 0

A zero-bonus test checks the other meaningful case: without a change to a level value, the original and recalculated totals stay equal.

```python
# CODE_RUNNER: UESL 3.1 Test - Zero bonus
level_one_stars = 12
level_two_stars = 18
level_three_stars = 10
bonus = 0

total_stars = level_one_stars + level_two_stars + level_three_stars
level_two_stars = level_two_stars + bonus
recalculated_total = level_one_stars + level_two_stars + level_three_stars

print("Original total:", total_stars)
print("Recalculated total:", recalculated_total)
assert (total_stars, recalculated_total) == (40, 40)
print("Validation: both totals remain 40 with no bonus.")
```

## Design Thinking

Some players may prefer less on-screen motion. This prototype stores `reduced_motion = True` as a clear on/off accessibility choice while keeping progress in separate level variables. I tested both a bonus of 5 and a bonus of 0 to show why a total must be explicitly recalculated after its inputs change.
