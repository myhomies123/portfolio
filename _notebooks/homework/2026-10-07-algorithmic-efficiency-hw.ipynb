## Popcorn
Prediction for n=8: 8 linear visits, 8 copy operations, 4 logarithmic steps. Changing n to 16 doubles linear work but adds only one logarithmic step. Counts model growth, not elapsed timing.


````python
# CODE_RUNNER: 2026-10-07-algorithmic-efficiency-hw exercise 3
import math
for n in [8, 16]:
    visits = 0
    copies = 0
    for index in range(n):
        visits += 1
        copies += 1
    logarithmic_steps = math.floor(math.log2(n)) + 1
    print(f"n={n}: visits={visits}, copies={copies}, binary bound={logarithmic_steps}")
    assert (visits, copies, logarithmic_steps) == ((8,8,4) if n == 8 else (16,16,5))
````

## Homework
For 12 catalog items, one pass visits 12, a copy stores 12 entries, and all ordered pairs count 144. Doubling to 24 gives 24,24,576: quadratic work quadruples. For 1,000 sorted entries, binary search requires at most 10 comparisons while linear search may require 1,000. Repeated lookups favor binary search. An unsorted catalog requires sorting first; for one query that preparation may not pay off. A copied sorted list also needs O(n) additional storage. Polynomial growth is generally more scalable than exponential growth; large hard optimization problems may justify a heuristic with a quality tradeoff.


````python
# CODE_RUNNER: 2026-10-07-algorithmic-efficiency-hw exercise 6
def catalog_growth(n):
    """Return explicit linear visits, copied entries, and ordered-pair count."""
    if n < 0:
        raise ValueError("Catalog size must be nonnegative")
    catalog = list(range(n))
    visits = 0
    for item in catalog:
        visits += 1
    copied = []
    for item in catalog:
        copied.append(item)
    pairs = 0
    for first in catalog:
        for second in catalog:
            pairs += 1
    return visits, len(copied), pairs
for n in [12, 24, 0]:
    visits, copied, pairs = catalog_growth(n)
    print(f"n={n}: visits={visits}, copies={copied}, pairs={pairs}")
assert catalog_growth(12) == (12,12,144)
assert catalog_growth(24) == (24,24,576)
assert catalog_growth(0) == (0,0,0)
````

## Tests
Main input 12 → (12,12,144); doubled input 24 → (24,24,576); empty input 0 → (0,0,0). The empty case also needs a guard before log2, which is undefined at zero. The following independent test verifies both search results and actual comparison counts.


````python
# CODE_RUNNER: 2026-10-07-algorithmic-efficiency-hw exercise 8
import math
def binary_bound(n):
    if n < 0:
        raise ValueError("Negative input")
    return 0 if n == 0 else math.floor(math.log2(n)) + 1
def linear_search(values, target):
    comparisons = 0
    for index in range(len(values)):
        comparisons += 1
        if values[index] == target:
            return index, comparisons
    return -1, comparisons
def binary_search(values, target):
    low, high, comparisons = 0, len(values)-1, 0
    while low <= high:
        middle = (low+high)//2
        comparisons += 1
        if values[middle] == target:
            return middle, comparisons
        elif values[middle] < target:
            low = middle+1
        else:
            high = middle-1
    return -1, comparisons
assert binary_bound(0) == 0 and binary_bound(1000) == 10
for values in [[], [3], [3,5,7,9,11,13,15,17]]:
    for target in [3, 11, 17, 99]:
        linear = linear_search(values, target)
        binary = binary_search(values, target)
        assert linear[0] == binary[0]
        assert binary[1] <= binary_bound(len(values))
        print(f"{values} target={target}: linear={linear}, binary={binary}")
print("Empty, found, missing, and comparison-bound tests passed.")
````

## Design Thinking
**Need:** a librarian wants quick catalog lookup without excessive storage. **Goal:** compare growth before selecting an implementation. **Alternatives:** sequential search works without sorting; binary search needs sorted data but scales better for repeated queries. **Prototype:** explicit operation counters and a logarithmic bound above. **Test and revise:** I added empty-input handling after identifying log2(0) as invalid, then checked zero, singleton, found, and missing searches. I also labeled the output to separate operations from time. **Decision:** maintain a sorted catalog for repeated searches; use a direct pass for tiny or one-off unsorted datasets.
